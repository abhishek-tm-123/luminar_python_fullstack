## dictionary

- Dictionary stores as key:value pairs
- Dictionary can be defined with {}
- values can be accessed using key


```python
expenses = {"mon":750,"tue":850,"wed":550,"thu":570,"fri":780,"sat":1000}

print(expenses["tue"])

print(expenses["sat"])
```

Adding and updating value

```python
calories = {"mon":1850,"tue":2000,"wed":1750,"thu":1900,"fri":1850,"sat":1700}

print(calories["thu"])
print(calories)
#adding new key value
calories["sun"] = 2000

print(calories)

#updating value

calories["tue"]+=100 # updating existing value

print(calories)
```

```python
item={"id":10,"title":"lp-shirt","category":"clothings","price":34000}

#display price
print(item["price"])

#add an offer of 100
item["offer"]=100

print(item)

#update price as current price + 100
item["price"]+=100

print(item)
```

### methods

- `keys()` returns all key in a dictionary
- `values()` returns all values in a dictionary
- `item()` return both key and values
- `pop(key)` remove an element from dictionary with that key
- `update({key:value,key:value})` update dictionary with new dictionary
- `get(key)` return value of the key

```python
item={"id":10,"title":"lp-shirt","category":"clothings","price":34000}


for k in item.keys():
    print(k,item[k])
```

- `values()` 

```python
item={"id":10,"title":"lp-shirt","category":"clothings","price":34000}

for v in item.values():
    print(v)
```

- `items()`

```python
item={"id":10,"title":"lp-shirt","category":"clothings","price":34000}

for k,v in item.items():

    print(k,v)

print(item.items())
```

- `pop(key)`

```python
item={"id":10,"title":"lp-shirt","category":"clothings","price":34000}

item.pop("id")

print(item)
```

```python
#movie details

movie_details = {"id":1,"title":"BKU","year":2026,"run_time":130,"director":"girish AD","language":"malayalam","rating":4.5}

print(movie_details)
```

- `update({key:value})`

```python
print(movie_details)

movie_details.update({"title":"vikram","year":2022,"director":"lokesh k"})

print(movie_details)
```

- `get(key)`

```python
print(movie_details)

print(movie_details.get("director"))

print(movie_details.get("actors")) # return none if key doesn't exist
```
