### self constructor

it is used to point the current object

````python
class Employee:
    id:int
    name:str
    department:str
    salary:int

    def set_attributes(self,id,name,department,salary):
        self.id = id
        self.name = name
        self.department = department
        self.salary = salary

    def display_doctor(self):
        print(self.name,self.department,self.salary)
````

````python
class Doctor:

    name:str
    specialization:str
    experience:int

    def set_attributes(self,name,specialization,experience):
        self.name = name
        self.specialization = specialization
        self.experience = experience

    def display_doctor(self):
        print(self.name,self.specialization,self.experience)

dr_instance1 = Doctor()
dr_instance1.set_attributes("abhi","ortho",4)
dr_instance1.display_doctor()
````

````python
class Book:
    id:int
    title:str
    author:str
    price:int
    pages:int

    def set_attribute(self,id,title,author,price,pages):
        self.id = id
        self.title = title
        self.author = author
        self.price = price
        self.pages = pages

    def display_book(self):
        print(self.id,self.title,self.author,self.price,self.pages)

book_instance1 = Book()
book_instance1.set_attribute(1,"randamoozham","mt",650,670)
book_instance1.display_book()
````

### construcor
---

```
constuctor is a special method that initialize attributes of a class
name of constructor will be __init__
constructor automatically invoked during object creation 
```

````python
class Book:
    id:int
    title:str
    author:str
    price:int
    pages:int

    def __init__(self,id,title,author,price,pages):
        self.id = id
        self.title = title
        self.author = author
        self.price = price
        self.pages = pages

    def display_book(self):
        print(self.id,self.title,self.author,self.price,self.pages)

book_instance1 = Book(1,"randamoozham","mt",650,670)

book_instance1.display_book()
````

````python
class Student:
    roll_no:int
    name:str
    course:str
    def __init__(self,roll_no,name,course):
        self.roll_no = roll_no
        self.name = name
        self.course = course

    def display_student(self):
        print(self.roll_no,self.name,self.course)

student_instance1 = Student(1,"abhishek","django")
student_instance1.display_student()
````

#### Inheritance

```
child class accessing attributes and methods of parent class
```

````python
class Parent:
    def home(self):
        print("1000 sft home")

class Child(Parent):
    def social_media_accounts(self):
        print("insta, youtube, ...")

child_instance1 = Child()
child_instance1.home()
child_instance1.social_media_accounts()
````

#### multiple inheritance

`child class accessing attributes and methods of multiple parent class `

````python
class Person:

    def eat(self):
        print("Eating ...")

    def walk(self):
        print("walking.....")

class CricketPlayer:

    def bat(self):
        print("bating ...")

    def bowl(self):
        print("bowling.....")

class Developer(Person,CricketPlayer):

    def code(self):
        print("codinging ...")

    def debug(self):
        print("debugging.....")

developer_instance = Developer()

developer_instance.code()
developer_instance.bat()
developer_instance.walk()
````
