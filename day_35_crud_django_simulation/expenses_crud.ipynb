```python
class Expenses:
    def __init__(self):
        self.expenses=[
            {"id":1,"title":"biriyani","category":"food","amount":350,"date":"10/10/2026"}
        ]

    def get(self):
        for di in self.expenses:
            print(di)

    def post(self,**kwargs):
        form_data = kwargs
        self.expenses.append(form_data)
        print("expense added")

    def retrieve(self,id=None):
        expense = [di for di in self.expenses if di.get("id") == id][0]
        print(expense)
    
    def put(self,id=None,**kwargs):
        form_data = kwargs
        expense = [di for di in self.expenses if di.get("id") == id][0]
        expense.update(form_data)
        print("expense updated",expense)

    def delete(self,id=None):
        expense = [di for di in self.expenses if di.get("id") == id][0]
        self.expenses.remove(expense)
        print("expense deleted")

    

        
```

```python
expense_instance = Expenses()
expense_instance.get()
```

```python
expense_instance.post(id=2,title="meal",category="food",amount=200,date="11/10/2026")
```

```python
expense_instance.retrieve(id=2)
```

```python
expense_instance.put(id=2,amount=100)
```

```python
expense_instance.delete(id=2)
```
