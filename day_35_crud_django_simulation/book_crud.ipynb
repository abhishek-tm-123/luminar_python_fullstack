# book crud using class


get - list data  
post - add/create data  
retrieve - list one data  
put - update data  
delete - delete data  

```python
class Book:

    def __init__(self):
        self.book_records=[
            {"id":1,"title":"randamoozham","auther":"mt","price":550,"pages":550}
        ]

    def get(self):
        for di in self.book_records:
            print(di)

    def post(self,**kwargs):
        form_data = kwargs
        self.book_records.append(form_data)
        print("book inserted .......")

    def retrieve(self,id=None):
        book = [di for di in self.book_records if di.get("id") == id][0]
        print(book)

    def put(self,id=None,**kwargs):
        form_data = kwargs
        book = [di for di in self.book_records if di.get("id") == id][0]
        book.update(form_data)
        print("book updated",book)

    def delete(self,id=None):
        book = [di for di in self.book_records if di.get("id") == id][0]
        self.book_records.remove(book)
        print("book removed")


```

```python
book_instance = Book()
book_instance.post(id=2,title="aadujeevitham",auther="benyamin",price=650,pages=550)
#book_instance.get()
```

```python
book_instance.retrieve(id=1)
```

```python
book_instance.put(id=1,price=1200)
```

```python
book_instance.delete(id=2)
```

```python
book_instance.get()
```

```python
class FoodLog:
    def __init__(self):
        self.food_log_records = [
            {"id":1,"name":"idli","meal_type":"breakfast","calorie":200,"serving_size":2}
        ]

    def post(self,**kwargs):
        form_data = kwargs
        self.food_log_records.append(form_data)
        print("food log added ....")

    def get(self):
        for di in self.food_log_records:
            print(di)

    def retrieve(self,id=None):
        food_log = [di for di in self.food_log_records if di.get("id") == id][0]
        print(food_log)

    def put(self,id=None,**kwargs):
        form_data = kwargs
        food_log = [di for di in self.food_log_records if di.get("id")==id][0]
        food_log.update(form_data)
        print("food log updated",food_log)

    def delete(self,id=None):
        food_log = [di for di in self.food_log_records if di.get("id")==id][0]
        self.food_log_records.remove(food_log)
        print("food log removed")
```

```python
food_log_instance = FoodLog()
food_log_instance.get()
```

```python
food_log_instance.post(id=2,name="dosa",meal_type="breakfast",calorie=300,serving_size=2)
```

```python
food_log_instance.put(id=2,calorie=250,serving_size=3)
```

```python
food_log_instance.get()
```

```python
food_log_instance.delete(id=2)
```
