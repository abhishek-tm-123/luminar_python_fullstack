```python
class Vehicle:
    def __init__(self):
        self.vehicles=[
            {"id":1,"title":"CB 350","color":"red","brand":"honda","price":200000}
        ]

    def get(self):
        for di in self.vehicles:
            print(di)

    def post(self,**kwargs):
        form_data = kwargs
        self.vehicles.append(form_data)
        print("vehicle added")

    def retrieve(self,id=None):
        vehicle = [di for di in self.vehicles if di.get("id") == id][0]
        print(vehicle)
    
    def put(self,id=None,**kwargs):
        form_data = kwargs
        vehicle = [di for di in self.vehicles if di.get("id") == id][0]
        vehicle.update(form_data)
        print("vehicle updated",vehicle)

    def delete(self,id=None):
        vehicle = [di for di in self.vehicles if di.get("id") == id][0]
        self.vehicles.remove(vehicle)
        print("vehicle deleted")

    

        
```

```python
vehicle_instance = Vehicle()

vehicle_instance.get()
```

```python
vehicle_instance.post(id=2,title="duke",color="orange",brand="ktm",rating=250000)
```

```python
vehicle_instance.retrieve(id=2)
```

```python
vehicle_instance.put(id=2,rating=275000)
```

```python
vehicle_instance.delete(id=2)
```
