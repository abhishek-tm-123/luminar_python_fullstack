## movies crud

```python
class Movies:
    def __init__(self):
        self.movies=[
            {"id":1,"title":"kgf","year":2018,"run_time":160,"rating":8}
        ]

    def get(self):
        for di in self.movies:
            print(di)

    def post(self,**kwargs):
        form_data = kwargs
        self.movies.append(form_data)
        print("movie added")

    def retrieve(self,id=None):
        movie = [di for di in self.movies if di.get("id") == id][0]
        print(movie)
    
    def put(self,id=None,**kwargs):
        form_data = kwargs
        movie = [di for di in self.movies if di.get("id") == id][0]
        movie.update(form_data)
        print("movie updated",movie)

    def delete(self,id=None):
        movie = [di for di in self.movies if di.get("id") == id][0]
        self.movies.remove(movie)
        print("movie deleted")

    

        
```

```python
movie_instance = Movies()

movie_instance.get()
```

```python
movie_instance.post(id=2,title="kgf2",year=2022,run_time=160,rating=9)
```

```python
movie_instance.retrieve(id=2)
```

```python
movie_instance.put(id=2,run_time=170,rating=9.5)
```

```python
movie_instance.delete(id=2)
```
