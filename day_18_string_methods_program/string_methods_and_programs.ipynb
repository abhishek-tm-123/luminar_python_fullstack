## index(value)
`index(value) => returns index of first occurense of value`  
Eg:  

```
word="abhi"
word.index(b) => 1
```

````python
text="hello"

pos = text.index("l")

print(pos)

````

## find(value)  

`find(value) => return index of first occurance of value return -1 if value not exist in str`

```
text = "hello"

pos = text.find("l")=>3
```

````python
text = "hello"

print(text.find("l"))
````

## rfind(value)

```
rfind(value) => return index position of last occurence return -1 if value not exist in str

"hello".rfind("l") => 3
```

````python
print("hello".rfind("l")) #=> 3
````

## replace(old,new,count)  

`replace old value with new value`  

`count => for how many elements should replace with new value`

````python
text = "Hello"

new_text = text.replace("l","L")

print(new_text)
````
