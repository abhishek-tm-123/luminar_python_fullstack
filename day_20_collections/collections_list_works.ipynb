### list

` create => variable_name = [] eg. nums = [10,20,30]`

````python
fvt_colors = ["red","green","blue"]

print(fvt_colors[0])

print(fvt_colors[1])

print(fvt_colors)
````

````python
expenses = [1200,22000,456,6544]

for i in range(0,4):
    print(expenses[i])
````

````python
for e in expenses:
    print(e)
````

````python
daily_calories = [1800,2000,1600,1700,1600,2000]

for i in range(0,len(daily_calories)):
    print(daily_calories[i])
````

````python
for cal in daily_calories:
    print(cal)
````

````python
# update list

daily_calories[0] = 1900

print(daily_calories)
````

### list method


- `append(value) => appends value to end of the list`

````python
daily_calories.append(2100)
print(daily_calories)
````

````python
arr = [10,11,12,13,14,15,16,17]

evens=[]

for num in arr:

    if num%2==0:

        evens.append(num)

print(evens)
````

````python
arr = [10,11,12,13,14,15,16,17]

odds=[]

for num in arr:

    if num%2!=0:

        odds.append(num)

print(odds)
````

````python
scores = [45,56,67,65,34,45,40,20,78,90]

passed = []

failed = []

for mark in scores:

    if mark>=50:
        passed.append(mark)
    else:
        failed.append(mark)

print("passed  :",passed)

print("failed :",failed)
````

- `insert(pos,value) => insert value to the specified position of list`

````python
fvt_colors = ["red","green","blue"]

fvt_colors.insert(0,"black")

print(fvt_colors)
````

- `pop(index=-1)=> remove the value at specified index default =-1`

````python
arr = [10,20,30,40]

arr.pop(2)

print(arr)
````
