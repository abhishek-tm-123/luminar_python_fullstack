## set

` unordered , unique and mutable`

- `define set with st=set() or st={10,20,30}`

```python
st = {10,20,30,40,50,40,30}

print(st)
```

### methods 

----

- add(value)

```python
st = {10,20,30,40,50,40,30}
print(st)

st.add(100)

print(st)
```

- union(sequence)
- intersection(sequence)
- difference(sequence)

```python
st1 = {10,20,30,40,50,60}

st2 = {10,20,100,200,300,400}

union_set = st1.union(st2)
print(union_set)

intersection_set = st1.intersection(st2)
print(intersection_set)

differnce_set = st1.difference(st2)
print(differnce_set)
```

- issuperset(set)
- issubset(set)



```python
setA = {10,20,30,40,50,60}

setB = {30,40,50,60}

print(setA.issuperset(setB))

print(setB.issubset(setA))
```
