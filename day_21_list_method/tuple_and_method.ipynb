## tuple

`ordered , allow duplicates but immutable`

---

- `tuple can define with ()`

```python
courses = ("django","mearn" , "testing")

courses[0] = "data science" #error because tuple immutable
```

### tuple methods

---



### index(value)

`return index of value from tuple`

```python
courses = ("django","mearn" , "testing","testing")

print(courses.index("django"))
```

### count(value)

`return frequency of value from tuple`

```python
print(courses.count("testing"))
```
