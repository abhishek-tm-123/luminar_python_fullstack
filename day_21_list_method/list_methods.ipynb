### index(value)

`returns index of value`

```python
arr = [10,20,30,40,50]

print(arr.index(20))
```

### remove(value)

` remove the first occurence of value from the list`

```python
arr = [10,20,30,40,50,30]

arr.remove(30)

print(arr)
```

### count(value)

`return frequency of value from list`

```python
arr = [10,20,40,50,10,40,30,40,10,50,10,10]

print(arr.count(10))
```

### sort()

` sort list `

```python
#asce
arr = [1,5,2,3,6,7,8,4]
arr.sort()
print(arr)
```

```python
#desc
arr.sort(reverse=True)
print(arr)
```

### reverse()

`reverse the list`


```python
chars = ["t","a","n"]

chars.reverse()

print(chars)
```
