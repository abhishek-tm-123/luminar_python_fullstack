## Error Handling

handling errors that occur while a program is running

````python
num = num = int(input("Enter a number :"))
print(10/num)
print("hello")
````

### handling error

````python
try:
    num = int(input("Enter a number :"))
    print(10/num)
except ZeroDivisionError:
    print("Cannot divide by zero")
print("hello")
````

### syntax

```
try:
    code may cause error
except:
    handling error
else:
    run when there is no error
finally:
    execute when error occured or not

```

````python
try:
    num = int(input("Enter a number:"))
except ValueError:
    print("Enter a integer value")
else:
    print(num**2)
````

````python
#odd or even
try:
    num = int(input("enter a number :"))
except ValueError:
    print("enter a integer value")
else:
    if num%2==0:
        print(num,"is even")
    else:
        print(num,"is odd")
````

````python
# avg of two number

try:
    num1 = int(input("enter first number :"))
    num2 = int(input("enter second number:"))
except ValueError:
    print("Enter integer values")
else:
    total = num1+num2
    print("avg:",total/2)
````

````python
# percentage calculator

try:
    obtained_mark = int(input("enter your obtained mark :"))
    total_mark = int(input("enter total mark :"))
except ValueError:
    print("Enter integer value")
else:
    percentage = (obtained_mark/total_mark)*100
    print(percentage)
````

finally: execute when error occur or not

use: in bank transcation at last the finally block works to close the transaction

````python
# result

try:
    mark = int(input("Enter your mark :"))
except ValueError:
    print("Enter integer value")
else:
    if mark>50:
        print("pass")
    else:
        print("fail")
finally:
    print("Result published")
````
