### multilevel inheritance

` child class inherit from parent class and parent class inheriting from grandparent class`  
` child class accessing grand parent class properties and methods`

```python
class GrandParent:

    def properties(self):
        print(" 50 cent land")

class Parent(GrandParent):
    def home(self):
        print(" 1000sft home")

class Child(Parent):
    def social_media_accounts(self):
        print("social media accounts")

child_instance = Child()

child_instance.social_media_accounts()
child_instance.home()
child_instance.properties()
```

### Polymorphism ( more than one form) => same name different behavior

- method overloading => same method name different number of parameter
- method overriding => child redifining methods that are already defined in parent class

```python
class Calculator:
    def add(self,num1,num2):
        print(num1+num2)

    def add(self,num1,num2,num3):
            print(num1+num2+num3)

    def add(self,num1,num2,num3,num4):
            print(num1+num2+num3+num4)

calc_instance = Calculator()

calc_instance.add(10,29,30,40)

calc_instance.add(10,20)

```

```python
#method overriding
class Parent:
    def mobile(self):
        print("redmi note 14 pro")

    def bike(self):
        print("passion pro")

class Child(Parent):

    def mobile(self):
        print("iphone 18 pro")
    
    
    def bike(self):
        print("Honda CB 350")

child_instance = Child()

child_instance.mobile()
child_instance.bike()
```

If the parent class have constructor ,then it is child class's responsibility is to call parent classes constructor from the child class constuctor

```python
class Shape:
    name:str

    def __init__(self,name):
        self.name = name

class Square(Shape):

    side : int

    def __init__(self, name,side):
        super().__init__(name)  # calling parent class's constructor
        self.side = side

    def area(self):
        print("area of ",self.name,self.side**2 )

square_instance = Square("Square",12)
square_instance.area()
```

```python
class Rectangle(Shape):
    l:int
    w:int

    def __init__(self, name,l,w):
        super().__init__(name)
        self.l = l
        self.w = w

    def area(self):
        print("area of",self.name,self.l * self.w)

rec_instance = Rectangle("Rectangle",10,20)
rec_instance.area()
        
```

### Abstaction (hiding implementation details)

`showing only essential details`

when the child class inheriting from abstarct parent class it should implement all abstarct methods

ABC => Abstract Base Class


```python
class Triangle(Shape):
    b:int
    h:int

    def __init__(self, name,l,b):
        super().__init__(name)
        self.l = l
        self.b = b

    def area(self):
        print("area of",self.name,(1/2)*self.b*self.l)

triangle_instance = Triangle("traingle",10,20)
triangle_instance.area()
```

```python
class Circle(Shape):
    r:int

    def __init__(self, name,r):
        super().__init__(name)
        self.r = r

    def area(self):
        print("area of",self.name,3.14*(self.r**2))

circle_instance = Circle("circle",10)
circle_instance.area()
```

```python
class Parallelogram(Shape):
    b:int
    h:int

    def __init__(self, name,b,h):
        super().__init__(name)
        self.b=b
        self.h=h

    def area(self):
        print("area of",self.name,self.b*self.h)

parallelogram_instance=Parallelogram("parallelogram",10,20)
parallelogram_instance.area()
```

```python
class Trapezium(Shape):
    a:int
    b:int
    h:int

    def __init__(self, name,a,b,h):
        super().__init__(name)
        self.a=a
        self.b=b
        self.h=h

    def area(self):
        print("area of",self.name,(1/2)*(self.a*self.b)*self.h)

trapezium_instance=Trapezium("trapezium",10,20,30)
trapezium_instance.area()
```

```python
from abc import ABC,abstractmethod
class Editor(ABC):
    @abstractmethod
    def open(self):pass

    @abstractmethod
    def debug(self):pass

    @abstractmethod
    def execute(self):pass

class PyCharm(Editor):
    def open(self):
        print("pyc opening..")

    def debug(self):
        print("pyc debugging..")

    def execute(self):
        print("pyc executing")

pyc_instance = PyCharm()
pyc_instance.execute()
```
