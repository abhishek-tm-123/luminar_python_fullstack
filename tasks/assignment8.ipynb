1. SUM AND AVERAGE
   - Problem: Given a list of numerical values (integers or floats), write a 
     function that calculates and returns both the total sum and average value.
   - Input Example:  [10, 20, 30, 40, 50]
   - Output Example: Sum = 150, Average = 30.0
   - Constraint/Hint: Practice iterating through the list manually using a loop 
     before using standard library shortcuts.
     

```python


def sum_avg(arr):
    sum=0
    avg=0
    for num in arr:

        sum=sum+num

    avg = sum / len(arr)

    print("Sum =",sum)
    print("Average =",avg)

arr1 = [10,20,30,40,50]

sum_avg(arr1)
```

2. MIN & MAX FINDER
   - Problem: Find the largest and smallest numbers in a list without using 
     built-in functions like min() or max().
   - Input Example:  [3, 8, 1, 9, 4, 2]
   - Output Example: Min = 1, Max = 9
   - Constraint/Hint: Initialize min/max trackers with the first element of the list.

```python
arr = [3,8,1,9,4,2]
max = arr[0]
min = arr[0]

for num in arr:
    if num>max:
        max=num

    if num<min:
        min=num

print("Min = ",min)
print("Max = ",max)
```

5. LIST INVERSION (REVERSE)
   - Problem: Reverse a list in-place (or construct a reversed version) without 
     using built-in library methods (e.g., .reverse() or [::-1]).
   - Input Example:  [1, 2, 3, 4, 5]
   - Output Example: [5, 4, 3, 2, 1]
   - Constraint/Hint: Try using a two-pointer approach (swapping start and end elements).

```python
arr = [1,2,3,4,5]

start = 0
end = len(arr)-1

while start<end:
    temp=arr[start]
    arr[start]=arr[end]
    arr[end]=temp

    start+=1
    end-=1

print(arr)
```

11. MERGE TWO SORTED LISTS
    - Problem: Combine two pre-sorted lists into a single, fully sorted list.
    - Input Example:  List A = [1, 3, 5], List B = [2, 4, 6]
    - Output Example: [1, 2, 3, 4, 5, 6]
    - Constraint/Hint: Compare elements from both lists step-by-step using pointers.

```python
List_A = [1, 3, 5,7]
List_B = [2, 4, 6,8]

merged_list = []

i=0
j=0

while i<len(List_A) and j<len(List_B):
    if List_A[i]<List_B[j]:
        merged_list.append(List_A[i])
        i+=1
    else:
        merged_list.append(List_B[j])
        j+=1

while i<len(List_A):
    merged_list.append(List_A[i])
    i+=1

while j<len(List_B):
    merged_list.append(List_B[j])
    j+=1
            

print(merged_list)
```

13. LIST SPLITTING
    - Problem: Write a function that splits a single list into two halves. If 
      the list length is odd, the first half should take the extra element.
    - Input Example:  [10, 20, 30, 40, 50]
    - Output Example: Half 1 = [10, 20, 30], Half 2 = [40, 50]
    - Constraint/Hint: Calculate the midpoint index using integer division.


```python
def split(arr):
    mid = (len(arr) + 1) // 2

    first_half = arr[:mid]
    second_half = arr[mid:]
    return first_half, second_half

print(split([10,20,30,40,50]))
```

14. LIST ROTATION
    - Problem: Rotate elements of a list to the right by k positions.
    - Input Example:  List = [1, 2, 3, 4, 5], k = 2
    - Output Example: [4, 5, 1, 2, 3]
    - Constraint/Hint: Use slicing or pop/insert loops. Handle cases where k > length.

```python
arr = [1,2,3,4,5]
k = 2

def rotation(arr,k):
    k=k%len(arr)
    result = arr[-k:]+arr[:-k]
    return result

print(rotation(arr,k))
```

15. MOVE ZEROS TO END
    - Problem: Shift all zero values in an integer list to the end while 
      maintaining the relative order of non-zero elements.
    - Input Example:  [0, 1, 0, 3, 12]
    - Output Example: [1, 3, 12, 0, 0]
    - Constraint/Hint: Keep a index pointer for non-zero insertion.


```python
def shift_zero(arr):
    index = 0

    for i in range(len(arr)):
        if arr[i] != 0:

            temp = arr[index]
            arr[index] = arr[i]
            arr[i] = temp

            index += 1

    return arr

print(shift_zero([0, 1, 0, 3, 12]))
```

16. PALINDROME CHECKER
    - Problem: Determine whether a list reads the same forwards and backwards.
    - Input Example:  [1, 2, 3, 2, 1]
    - Output Example: True
    - Constraint/Hint: Compare element at index i with index (length - 1 - i).

```python
def is_palindrome(arr):
    n = len(arr)

    for i in range(n // 2):
        if arr[i] != arr[n - 1 - i]:
            return False

    return True


print(is_palindrome([1, 2, 3, 2, 1]))
```

17. SECOND LARGEST NUMBER
    - Problem: Find the second largest number in an unsorted list without 
      sorting the entire list.
    - Input Example:  [12, 35, 1, 10, 34, 1]
    - Output Example: 34
    - Constraint/Hint: Track two variables: largest and second_largest.

```python
def second_largest(arr):
    largest = arr[0]
    second_largest = arr[0]

    for num in arr:
        if num > largest:
            second_largest = largest
            largest = num

        elif num > second_largest and num != largest:
            second_largest = num

    return second_largest


print(second_largest([12, 35, 1, 10, 34, 1]))
```

18. AVERAGE WORD LENGTH
    - Problem: Given a list of string words, calculate the average character 
      length of the words in the list.
    - Input Example:  ["cat", "banana", "elephant", "dog"]
    - Output Example: 5.0  (Total characters = 20, Words = 4)
    - Constraint/Hint: Sum the length of each string and divide by total words count.

```python
def average_word_length(words):
    total = 0

    for word in words:
        total += len(word)

    average = total / len(words)

    return average


words = ["cat", "banana", "elephant", "dog"]

print(average_word_length(words))
```

20. RUN-LENGTH ENCODER
    - Problem: Compress consecutive repeating elements in a list into pairs 
      of (element, consecutive_count).
    - Input Example:  ['a', 'a', 'a', 'b', 'b', 'c', 'a', 'a']
    - Output Example: [('a', 3), ('b', 2), ('c', 1), ('a', 2)]
    - Constraint/Hint: Iterate while tracking the current element and its count.

```python
def run_length(arr):
    result = []

    current = arr[0]
    count = 1

    for i in range(1, len(arr)):
        if arr[i] == current:
            count += 1
        else:
            result.append((current, count))

            current = arr[i]
            count = 1

    
    result.append((current, count))

    return result


arr = ['a', 'a', 'a', 'b', 'b', 'c', 'a', 'a']

print(run_length(arr))
```
