# Object Oriented programming

**is a way of programming using classes and objects that allows us to bring real world entities to programming objects**

- class : **plan | design | template or blueprint of creating an object**
- object : **instance of class**

syntax:  
```
class Class name:
    attributes : it defines an object . it is the properties of an object
    methods : functionalities of an object can do 
```
