````python
# display even number from numberts.txt

fr = open("numbers.txt","r")

for line in fr:
    line = line.rstrip("\n")
    num = int(line)
    if num%2==0:
        print(num)
fr.close()
````

````python
# display odd number from numberts.txt

fr = open("numbers.txt","r")

for line in fr:
    line = line.rstrip("\n")
    num = int(line)
    if num%2!=0:
        print(num)
fr.close()
````

````python
# read from words.txt print palindrome words

fr = open("words.txt","r")

for line in fr:
    word = line.rstrip("\n")
    if word == word[::-1]:
        print(word)

fr.close()
````

````python
# find common prefix from words.txt

fr = open("words.txt","r")
all_words = []
for line in fr:
    word = line.rstrip("\n")
    all_words.append(word)
fr.close()

all_words.sort()

first = all_words[0]

last = all_words[-1]

prefix = ""

small_word = first if len(first)<len(last) else last

for i in range(len(small_word)):
    if first[i] == last[i]:
        prefix+=first[i]
    else:
        break

print(prefix)
````

````python
#orders count

all_orders = []

fr=open("orders.txt","r")
for line in fr:
    item = line.rstrip("\n")
    all_orders.append(item)

order_count = {o:all_orders.count(o) for o in all_orders}
print(order_count)

fr.close()
````

````python
fr = open("weather.txt","r")

weather_report = []

for line in fr:
    line = line.rstrip("\n")
    district,temp = line.split(",")
    dictionary = {"district":district,"temp":temp}
    weather_report.append(dictionary)

fr.close()
print(weather_report)
````

````python
fr = open("films.txt","r")
films_report = []

for line in fr:
    line=line.rstrip("\n")
    title,language,rating = line.split(",")
    dictionary = {"title":title,"language":language,"rating":rating}
    films_report.append(dictionary)

fr.close()

print(films_report)
````
