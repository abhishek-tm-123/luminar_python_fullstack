expense crud


```python
expenses =[
    {"id":1,"title":"biriyani","category":"food","amount":350,"date":"10/10/2026"}
]
```

add expense

```python
def add_expense(**kwargs):
    form_data = kwargs
    expenses.append(form_data)
    print("record inserted")

add_expense(id=2,title="petrol",category="fuel",amount=500,date="8/10/2026")
```

list all expenses

```python
def list_all_expenses():
    for di in expenses:
        print(di)

list_all_expenses()
```

list specific expense

```python
def retrieve_expense(id=None):
    exp_data = [di for di in expenses if di.get("id")==id][0]
    print(exp_data)

retrieve_expense(2)
```

delete expense

```python
def delete_expense(id=None):
    exp_data = [di for di in expenses if di.get("id") == id][0]
    expenses.remove(exp_data)
    print("expense removed")

delete_expense(2)
```

update expense

```python
def update_expense(id=None,**kwargs):
    exp_data = [di for di in expenses if di.get("id") == id][0]
    form_data = kwargs
    print(exp_data)
    exp_data.update(form_data)
    print(exp_data)

update_expense(1,amount=200,date="8/10/2026")
```
