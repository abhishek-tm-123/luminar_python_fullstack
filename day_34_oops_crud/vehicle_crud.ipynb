```python
vehicles=[
    {"id":1,"title":"CB 350","color":"red","brand":"honda","price":200000}
]
```

Create


```python
def add_vehicle(**kwargs):
    form_data = kwargs
    vehicles.append(form_data)
    print("record inserted")

add_vehicle(id=2,title="duke",color="orange",brand="ktm",rating=250000)
vehicles
```

list all movies

```python
def list_all_vehicles():
    for di in vehicles:
        print(di)

list_all_vehicles()
```

list specific movie

```python
def retrive_vehicle(id=None):
    vehicle_data = [di for di in vehicles if di.get("id") == id][0]
    print(vehicle_data)

retrive_vehicle(2)
```

delete a movie

```python
def delete_vehicle(id=None):
    vehicle_data = [di for di in vehicles if di.get("id") == id][0]

    vehicles.remove(vehicle_data)

    print("vehicle removed")

delete_vehicle(2)
```

```python
list_all_vehicles()
```

update movie

```python
def update_vehicle(id,**kwargs):
    form_data = kwargs
    vehicle_data = [di for di in vehicles if di.get("id") == id][0]
    print(vehicle_data)
    vehicle_data.update(form_data)
    print(vehicle_data)

update_vehicle(1,title = "rc",price=175000)
```
