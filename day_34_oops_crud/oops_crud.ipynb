```python
# student : id ,name,age,course,email

students =[
    {"id":1,"name":"rahul","age":20,"course":"django","email":"rahul@example.com"}
    
]
```

```python
#new student data
data = {"id":2,"name":"rohit","age":22,"course":"mearn","email":"rohit@example.com"}

students.append(data)

students
```

```python
#display all student

for di in students:
    print(di)
```

```python
# fetch one student

id = 1

student_data = [di for di in students if di.get("id")==id][0]

print(student_data)
```

```python
#delete student data

id =2
student_data = [di for di in students if di.get("id")==id][0]

students.remove(student_data)

print(students)
```

```python
#update student

form_data = {"age":23,"course":"testing","email":"test@gmail.com"}

id = 1

student_data = [di for di in students if di.get("id") == id][0]

print("before updating",student_data)

student_data.update(form_data)

print("after update",student_data)

print(students)
```

### Employee

```python
# employees = [id,name,department,slary,location]
employees = [
    {"id":1,"name":"abhi","department":"hr","salary":40000,"location":"tsr"}

]
```

add employee

```python
# add new employee

def add_employee(**kwargs):
    form_data = kwargs
    employees.append(form_data)
    print("record inserted")

add_employee(id=2,name="sam",department="sales",salary=50000,location="tvm")
```

**kwargs => make the function arguments in this form "id=1" to dictionary eg; {"id":1}

list all emplyees

```python
def list_all_employees():
    for di in employees:
        print(di)

list_all_employees()
```

list specific employee

```python
def retrive_employee(id=None):
    emp_data = [di for di in employees if di.get("id")==id][0]
    print(emp_data)

retrive_employee(id=2)
```

delete employee

```python
def delete_employee(id=None):
    emp_data = [di for di in employees if di.get("id")==id][0]
    employees.remove(emp_data)
    print("employee removed")

delete_employee(id=2)
```

```python
def update_employee(id=None,**kwargs):
    form_data = kwargs
    emp_data = [di for di in employees if di.get("id")==id][0]

    emp_data.update(form_data)
    print(emp_data)

update_employee(1,department = "sales",location = "ckd")
```
