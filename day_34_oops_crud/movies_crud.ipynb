```python
movies=[
    {"id":1,"title":"kgf","year":2018,"run_time":160,"rating":8}
]
```

Create


```python
def add_movie(**kwargs):
    form_data = kwargs
    movies.append(form_data)
    print("record inserted")

add_movie(id=2,title="kgf2",year=2022,run_time=172,rating=9)
movies
```

list all movies

```python
def list_all_movies():
    for di in movies:
        print(di)

list_all_movies()
```

list specific movie

```python
def retrive_movie(id=None):
    movie_data = [di for di in movies if di.get("id") == id][0]
    print(movie_data)

retrive_movie(2)
```

delete a movie

```python
def delete_movie(id=None):
    movie_data = [di for di in movies if di.get("id") == id][0]

    movies.remove(movie_data)

    print("movie removed")

delete_movie(2)
```

```python
list_all_movies()
```

update movie

```python
def update_movie(id,**kwargs):
    form_data = kwargs
    movie_data = [di for di in movies if di.get("id") == id][0]
    print(movie_data)
    movie_data.update(form_data)
    print(movie_data)

update_movie(1,run_time = 170,rating=9)
```
