# String slicing programs

> slicing extract a portion of string

```
eg:

text="python programing"
      01234567890123456

text[0:6] => python

text[:6] => python

text[7:] => programming
```

````python
text = "i like python and django"
#       012345678901234567901234
portion = text[0:13]

print(portion)

portion2 = text[18:]

print(portion2)
````

````python
word = "programming"

word_copy = word[:]

print(word_copy)
````

````python
word = "hello"

rev = word[::-1]

print(rev)
````

````python
# create a function to display digits in a given string

def display_digits(text):

    for char in text:

        if char.isdigit():
            print(char)

display_digits("i hvae 1 bike and 4 car")
````

````python
# create a function to display vowels count in a string

def vowel_count(text):

    vowels = "aeiou"
    v_count = 0

    for ch in text:

        if ch in vowels:

            v_count +=1
    print(v_count)

vowel_count("hello")
vowel_count("programming")
````

### anagram

silent => listen

tan => ant

two string contain same charecters with different order

logic:

first sort two string 

then check they are same

if true anagram 

if false not anagram

````python
text1 = "abc"
text2 = "bca"

sorted_text1 = sorted(text1)
sorted_text2 = sorted(text2)

print(sorted_text1==sorted_text2)
````

````python

def anagram(text1,text2):
    sorted_text1 = sorted(text1)
    sorted_text2 = sorted(text2)

    if sorted_text1 == sorted_text2 :

        print("words are anagram")

    else:

        print("word are not anagram")


anagram("abc","bca")
anagram("silent","listen")
````

## pangram

````python
alphabets = "abcdefghijklmnopqrstuvwxyz"

word = "the quick brown fox jumps over the lazy dog"

result = True

for ch in alphabets:
    if ch not in word :
        result=False
        break

print(result)
````
