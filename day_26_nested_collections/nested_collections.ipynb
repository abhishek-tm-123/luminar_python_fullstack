```python
#movies with higest rating

all_rating = [di.get("rating") for di in movies]

max_rating = max(all_rating)

top_movies = [di.get("title") for di in movies if di.get("rating") == max_rating]

print(top_movies)
```

```python
#movies with lowest rating

all_rating = [di.get("rating") for di in movies]

min_rating = min(all_rating)

low_movies = [di.get("title") for di in movies if di.get("rating") == min_rating]

print(low_movies)
```

```python
# movie with highest run_time

all_run_time = [di.get("run_time") for di in movies]

max_run_time = max(all_run_time)

movies_high_run_time = [di.get("title") for di in movies if di.get("run_time") == max_run_time]

print(movies_high_run_time)
```

```python
min_run_time = min(all_run_time)

movies_low_run_time = [di.get("title") for di in movies if di.get("run_time") == min_run_time]

print(movies_low_run_time)
```

```python
#movie with lowest rating

def get_rating(di):
    return di.get("rating")

print(min(movies,key=get_rating))
```

```python
#movie with highest rating

max(movies,key=get_rating)
```

```python
#movie with highest run_time

def get_run_time(di):
    return di.get("run_time")

max(movies,key=get_run_time)
```
