# comprehension

`easy way to creating list/set/dict from a sequecnce`

```
syntax : result = [return_value iteration condition]

```

EG:

```
nums = [1,2,3,4,5]

add_five = [num+5 for num in nums]
```

````python
nums = [10,11,12,13,14,15,16]

squares = [num**2 for num in nums]

print(squares)
````

````python
cubes = [num**3 for num in nums]

print(cubes)
````

````python
double = [num*2 for num in nums]
print(double)
````

````python
greater_30 = [num for num in nums if num>13]
print(greater_30)
````

````python
evens = [num for num in nums if num%2==0]

print(evens)
````

````python
words = ["flower","flight","fight","fly"]

upper_words = [w.upper() for w in words]

print(upper_words)
````

````python
gtr_3_words = [w for w in words if len(w)>3]

print(gtr_3_words)
````

````python
lst = [10,11,10,13,10,15,16,15,13]

counts = {num:lst.count(num) for num in lst}

print(counts)
````

````python
squares = {num:num**2 for num in lst}

print(squares)
````

````python
#common prefix

words = ["forward","forcast","forsee","for"]

words.sort()

first = words[0]

last = words[-1]

prefix = ""

small_word = first if len(first) < len(last) else last

for i in range (len(small_word)):
    if first[i] == last[i]:
        prefix+=first[i]

    else:
        break

print(prefix)
````
